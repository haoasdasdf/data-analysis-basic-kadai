```python


import matplotlib.pyplot as plt
import japanize_matplotlib

import pandas as pd

df = pd.read_csv('sample_pandas_6.csv')
df

```

```python
category_df = pd.read_csv('category.csv')
category_df
```

```python
df_mer = pd.merge(df, category_df[['商品番号', 'カテゴリー']], how='inner', on='商品番号')
df_mer
```

```python

df_mer['カテゴリー'].value_counts().plot(kind='bar')
plt.tight_layout()
plt.show()
```

```python
# df_mer.groupby('カテゴリー')['注文数'].size()
df_mer.groupby('商品番号')['注文数'].describe()
```
